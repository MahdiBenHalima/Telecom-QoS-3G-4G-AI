# Reproducible 3G QoS experiment
This notebook runs the reconstruction of the original 3G training protocol. It requires an authorized private `DT2.csv`. Targets are **derived from RSCP and Ec/N0**; metrics do not measure independent QoS ground truth.


```python
from pathlib import Path
from training.train_3g import load_3g_data, label_3g, train
DATA = Path('DT2.csv')  # replace with your authorized local dataset path
OUTPUT = Path('backend/models')
```

## Inspect and label data
Source data is not redistributed. Do not share notebook outputs containing operator measurements.


```python
if DATA.exists():
    frame = load_3g_data(DATA)
    print('Rows after cleaning:', len(frame))
    print('Derived label counts:', label_3g(frame).value_counts().to_dict())
else:
    print('Provide DT2.csv locally to run this notebook.')
```

## Reconstruct the original experiment
Stratified 80/20 split (`random_state=42`) compares Logistic Regression, Decision Tree and Random Forest; the final Random Forest is refitted on all rows and exported. See `docs/REPRODUCIBILITY.md`.


```python
if DATA.exists():
    comparisons = train(DATA, OUTPUT)
    for row in comparisons:
        print(row)
else:
    print('Training skipped: source dataset not present.')
```

## Interpretation
The prediction target is created from input KPIs. High accuracy reflects agreement with an engineered scoring policy, **not** external network QoS ground truth. Random splitting can overestimate deployment performance when repeated cell measurements are correlated.
